```python
from sklearn.datasets import make_moons
X, y = make_moons(n_samples=250, noise=0.25, random_state=1)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X[:,0], X[:,1], c=y)
plt.show()
```

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
from keras import Sequential
from keras.layers import Dense
from keras.callbacks import EarlyStopping
```

```python
from keras.optimizers import Adam, RMSprop, SGD, Adagrad
adam = Adam(learning_rate=0.01)
rms = RMSprop(learning_rate=0.01)
adagrad = Adagrad(learning_rate=0.01)
momentum = SGD(learning_rate=0.01, momentum=0.9)
nag = SGD(learning_rate=0.01, momentum=0.9, nesterov=True)
sgd = SGD(learning_rate=0.01)
```

```python
callback = EarlyStopping(monitor="val_loss", patience=3)
```

```python
histories = []
```

```python
model0 = Sequential()

model0.add(
    Dense(
        128, activation = "relu", input_dim = 2
    )
)
model0.add(
    Dense(
        128, activation = "relu"
    )
)
model0.add(
    Dense(
        1, activation = "sigmoid"
    )
)

model0.compile( loss = "binary_crossentropy", optimizer = adam, metrics = ["accuracy"])
model0.summary()
history = model0.fit(X, y, epochs = 100, batch_size = 10, callbacks = callback, validation_split=0.2, verbose = 1)

val_acc = history.history["val_accuracy"][-1]
val_loss = history.history["val_loss"][-1]
acc = history.history["accuracy"][-1]
loss = history.history["loss"][-1]
print(f" accuracy: {acc}\n val_accuracy: {val_acc}\n loss: {loss}\n val_loss: {val_loss}")
print("epochs: ", len(history.history["accuracy"]))

histories.append(history.history)
```
